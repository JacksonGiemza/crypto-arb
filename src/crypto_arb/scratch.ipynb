```python
import websocket
from decimal import Decimal
import json


url = "wss://advanced-trade-ws.coinbase.com"
subscription = {
    "type": "subscribe",
    "product_ids": ["BTC-USD","ETH-USD"],
    "channel": "level2"
}

ws = websocket.create_connection(url)
ws.send(json.dumps(subscription))



while True:
    message = ws.recv()
    data = json.loads(message)
    events = data["events"]
    for event in events:
        if data.get("channel") != "l2_data":
            continue
        print(
        data["sequence_num"],
        [event["product_id"] for event in data.get("events", [])]
        )
```
